## Data Cleaning

```python

print("Duplicates before cleaning:", df.duplicated().sum())


df = df.drop_duplicates()

print("Duplicates after cleaning:", df.duplicated().sum())
print("Dataset shape after removing duplicates:", df.shape)
```

```python


text_columns = [
    "Name",
    "Gender",
    "Department",
    "City",
    "Email",
    "Performance_Rating"
]

for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
# Standardize Department values

df["Department"] = df["Department"].str.lower()

df["Department"] = df["Department"].replace({
    "it": "IT",
    "hr": "HR",
    "finance": "Finance",
    "marketing": "Marketing",
    "sales": "Sales",
    "operations": "Operations",
    "data analytics": "Data Analytics"
})
```

```python
df["Department"].value_counts(dropna=False)
```

```python
# Standardize Gender values

df["Gender"] = df["Gender"].str.lower()

df["Gender"] = df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "other": "Other"
})
```

```python
df["Gender"].value_counts(dropna=False)
```

```python
# Standardize City values

df["City"] = df["City"].str.title()
```

```python
df["City"].value_counts(dropna=False)
```

```python
# Standardize Performance Rating values

df["Performance_Rating"] = (
    df["Performance_Rating"]
    .str.strip()
    .str.title()
)
```

```python
df["Performance_Rating"].value_counts()
```

```python
# Correct inconsistent Age values

df["Age"] = df["Age"].replace({
    "twenty-six": 26,
    "29 years": 29,
    "thirty-two": 32
})

# Convert Age to numeric
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df["Age"].dtype
```

```python
# Remove currency symbols and text from Salary

df["Salary"] = (
    df["Salary"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("Rs.", "", regex=False)
    .str.strip()
)

df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
```

```python
df["Salary"].dtype
```

```python
# Handle missing values

# Fill missing Age with median age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with median salary
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Fill missing Department with Unknown
df["Department"] = df["Department"].fillna("Unknown")

# Fill missing Joining Date with median date
median_date = df["Joining_Date"].dropna().median()
df["Joining_Date"] = df["Joining_Date"].fillna(median_date)
```

```python
# Convert Joining_Date to datetime

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)
```

```python
df["Joining_Date"].dtype
```

```python
print("========== FINAL DATA QUALITY CHECK ==========")

print("\nDataset shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
